# Satır alt kümeleri oluşturma

Veri biliminin büyük bir bölümü, veri kümenizin hangi kısımlarının ilginç olduğunu bulmakla ilgilidir. Bunun için en basit tekniklerden biri, belirli kriterlere uyan satırların bir alt kümesini bulmaktır. Bu işlem bazen _satırları filtreleme_ veya _satırları seçme_ olarak da bilinir.

Bir DataFrame’in alt kümesini oluşturmanın birçok yolu vardır. Belki de en yaygın olanı, her satır için `True` veya `False` döndürmek üzere ilişkisel işleçler kullanmak ve ardından köşeli ayraç içine aktarmaktır.

```
dogs[dogs["height_cm"] > 60]
dogs[dogs["color"] == "tan"]
```

“Bit düzeyinde and” (`&`) işlecini kullanarak birden fazla koşulu aynı anda filtreleyebilirsiniz.

```
dogs[(dogs["height_cm"] > 60) & (dogs["color"] == "tan")]
```
